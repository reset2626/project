```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.stats.diagnostic import acorr_ljungbox, het_arch

from arch import arch_model
```

# Problem 1：ARIMA 模型

根據題目要求，需要依序完成以下工作：

1. 檢驗資料是否具有單根。
2. 決定 ARIMA 模型的階數 $(p,i,q)$。
3. 估計模型參數並報告顯著性。
4. 檢查殘差是否為白噪音，以判斷模型是否合理。
5. 進行 1-step 到 5-step ahead forecasting，並報告信賴區間。

## 1. 載入資料並進行初步觀察

首先讀入 GDP 資料，並觀察其時間序列圖。  
透過視覺化可以先初步判斷資料是否具有趨勢、是否可能為非平穩序列。  
由於 ARIMA 模型通常需要對資料進行平穩化處理，因此這一步是後續分析的重要基礎。

```python
gdp_df = pd.read_csv(r"c:\Users\allen\OneDrive\Desktop\研究所\時序\US GDP.csv")
gdp_df.head()
```

```python
print(gdp_df.columns)
print(gdp_df.shape)
```

```python
# ===== 請依照你的資料格式修改這裡 =====
# 暫時先抓最後一欄作為 GDP 資料，若不是正確欄位請自行修改欄位名稱
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['Microsoft JhengHei']  # 微軟正黑體

gdp_series = gdp_df.iloc[:, -1]
gdp_series.index = pd.RangeIndex(start=1, stop=len(gdp_series)+1)

plt.figure(figsize=(10, 4))
plt.plot(gdp_series)
plt.title("US GDP 時間序列圖")
plt.xlabel("時間")
plt.ylabel("GDP")
plt.grid(True)
plt.show()
```

## 2. 單根檢定（Unit Root Test）

為了判斷 GDP 序列是否平穩，使用 Augmented Dickey-Fuller test 進行檢定。

檢定假設如下：

- 虛無假設 $H_0$：序列具有單根，亦即為非平穩序列。
- 對立假設 $H_1$：序列不具有單根，亦即為平穩序列。

若 p-value 大於 0.05，則無法拒絕虛無假設，表示資料可能具有單根，因此需要進一步差分。

```python
adf_result = adfuller(gdp_series.dropna())

print("ADF Statistic:", adf_result[0])
print("p-value:", adf_result[1])
print("使用的落後期數:", adf_result[2])
print("樣本數:", adf_result[3])
print("臨界值:")
for key, value in adf_result[4].items():
    print(f"   {key}: {value}")
```

### ADF 檢定結果說明

根據上述 ADF 檢定結果，可以判斷原始 GDP 序列是否存在單根。

- p-value 值為1，表示原始序列為非平穩序列。
- 此時需對資料進行差分，再重新進行 ADF 檢定。

ARIMA 模型中的差分次數 $i$，即是根據此步驟決定。

```python
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['Microsoft JhengHei']  # 微軟正黑體
plt.rcParams['axes.unicode_minus'] = False  # 解決負號顯示問題

gdp_diff1 = gdp_series.diff().dropna()

plt.figure(figsize=(10, 4))
plt.plot(gdp_diff1)
plt.title("GDP 一階差分後序列")
plt.xlabel("時間")
plt.ylabel("差分後 GDP")
plt.grid(True)
plt.show()
```

```python
adf_diff1 = adfuller(gdp_diff1)

print("ADF Statistic (一階差分):", adf_diff1[0])
print("p-value:", adf_diff1[1])
print("使用的落後期數:", adf_diff1[2])
print("樣本數:", adf_diff1[3])
print("臨界值:")
for key, value in adf_diff1[4].items():
    print(f"   {key}: {value}")
```

### (a) 單根檢定結果

本研究使用 Augmented Dickey-Fuller (ADF) 檢定來判斷 GDP 序列是否具有單根。

檢定結果顯示，一階差分後序列的 ADF statistic 為 -3.4959，對應的 p-value 為 0.008，小於顯著水準 0.05，因此拒絕「存在單根」的虛無假設。

由此可知，原始 GDP 序列為非平穩序列，但經過一階差分後已轉為平穩序列。因此，在 ARIMA 模型中選擇差分次數 $i = 1$。

```python
plot_acf(gdp_diff1, lags=20)
plt.show()

plot_pacf(gdp_diff1, lags=20, method='ywm')
plt.show()
```

```python
candidate_orders = [(1,1,0), (2,1,0), (3,1,0), (1,1,1), (2,1,1)]

results = []

for order in candidate_orders:
    try:
        model = ARIMA(gdp_series, order=order)
        fitted = model.fit()
        results.append({
            "order": order,
            "AIC": fitted.aic,
            "BIC": fitted.bic
        })
    except:
        pass

pd.DataFrame(results).sort_values("AIC")
```

### (b) ARIMA 模型階數選擇

根據一階差分後序列的 ACF 與 PACF 圖進行初步判斷。

由圖可觀察到：
- ACF 呈現緩慢衰減，顯示序列可能具有 AR 結構。
- PACF 在前幾個落後期（約 lag 3）出現顯著值，之後迅速趨近於 0，呈現截尾現象。

因此，初步選擇數個候選模型，包括 ARIMA(1,1,0)、ARIMA(2,1,0)、ARIMA(3,1,0)、ARIMA(1,1,1) 與 ARIMA(2,1,1)。

接著透過 AIC 與 BIC 進行模型比較。結果顯示，ARIMA(2,1,1) 在所有候選模型中具有最小的 AIC 與 BIC 值，因此選擇 ARIMA(2,1,1) 作為最終模型。

```python
arima_model = ARIMA(gdp_series, order=(2,1,1))
arima_fit = arima_model.fit()

print(arima_fit.summary())
```

### (c) 參數估計結果

由估計結果可得：

- AR(1) 係數為 0.8629（p-value = 0.000）
- AR(2) 係數為 0.1350（p-value = 0.000）
- MA(1) 係數為 -0.9003（p-value = 0.000）

所有參數在 5% 顯著水準下皆為統計顯著，表示模型中的 AR 與 MA 皆對資料具有顯著影響。

其中，AR(1) 與 AR(2) 顯示過去兩期的變動對目前 GDP 變化具有影響，而 MA(1) 則表示隨機衝擊項亦對當期變動具有顯著影響。

因此，ARIMA(2,1,1) 模型能有效描述 GDP 序列的動態結構。

```python
from statsmodels.graphics.tsaplots import plot_acf

resid = arima_fit.resid

plot_acf(resid.dropna(), lags=20)
plt.title("殘差 ACF")
plt.show()
```

### (d) 殘差診斷

為檢查 ARIMA(2,1,1) 模型是否合理，針對殘差進行白噪音檢定。

首先觀察殘差的 ACF 圖，可以發現大部分自相關係數皆落在信賴區間內，未出現顯著的自相關結構。

此外，根據 Ljung-Box 檢定結果，其 p-value 為 0.91，大於顯著水準 0.05，因此無法拒絕殘差為白噪音的虛無假設。

綜合上述結果，顯示殘差不具有顯著自相關，表示模型已充分捕捉資料中的時間相依結構，因此 ARIMA(2,1,1) 模型為合理模型。

```python
forecast_res = arima_fit.get_forecast(steps=5)

forecast_table = forecast_res.summary_frame()
forecast_table
```

### (e) 預測結果

根據已建立之 ARIMA(2,1,1) 模型，進行 1-step 至 5-step ahead 的預測，並同時提供信賴區間。

預測結果如下：

由預測結果可觀察到，未來 GDP 呈現逐步上升的趨勢，顯示模型預測經濟持續成長。

此外，隨著預測期數增加，信賴區間逐漸變寬，表示預測的不確定性隨時間增加，此為時間序列預測的常見現象。

由於前述殘差檢定結果顯示模型合理，因此本研究認為 ARIMA(2,1,1) 模型具有良好的預測能力。

# Problem 2：GARCH 模型

根據題目要求，本題使用 3M 股票的月報酬率。

首先需將原始資料轉換為對數報酬率（log-return）：

$Y_t = \log\left(\frac{P_t}{P_{t-1}}\right)$

本題的主要目的是建立 GARCH 模型來描述報酬率的波動性，並完成以下工作：

- 先把月報酬率轉成 `log-return`。
- 再用 `ARIMA` 配適平均方程。
- 對 `ARIMA` 殘差做 ARCH effect 檢定。
- 再對殘差建立 `GARCH(1,1)` 模型。
- `Y_t` 的預測由 ARIMA mean model 提供。
- volatility 的預測由 GARCH variance model 提供。

## 1. 資料前處理與 log-return 計算

先讀入 3M 股票資料，並將價格序列轉換為對數報酬率（log-return）。

題目指定 forecast origin 為 `2008-12-31`，因此後續的模型建立與預測皆以前述日期之前的資料作為樣本。

```python
m3_df = pd.read_csv(
    r"c:\Users\allen\OneDrive\Desktop\研究所\時序\Dataset_3M.txt",
    sep=r"\s+"
)

m3_df.columns = m3_df.columns.str.strip()
m3_df["date"] = pd.to_datetime(m3_df["date"].astype(str)).dt.to_period("M").dt.to_timestamp("M")
m3_df["rtn"] = pd.to_numeric(m3_df["rtn"], errors="coerce")
m3_df["Y"] = np.log1p(m3_df["rtn"])
m3_df = m3_df.dropna().set_index("date")

print(m3_df.head())
print(m3_df.columns.tolist())

forecast_origin = pd.Timestamp("2008-12-31")
Y_train = m3_df.loc[:forecast_origin, "Y"]

plt.figure(figsize=(10, 4))
plt.plot(Y_train)
plt.title("3M 股票 log-return（截至 2008-12-31）")
plt.xlabel("Date")
plt.ylabel("Y_t")
plt.grid(True)
plt.show()
```

```python
from statsmodels.stats.diagnostic import het_arch, acorr_ljungbox
from statsmodels.tsa.arima.model import ARIMA

candidate_orders = [(0,0,1), (1,0,0), (1,0,1), (2,0,0), (2,0,1), (2,0,2), (3,0,0)]
mean_results = []

for order in candidate_orders:
    try:
        fit = ARIMA(Y_train, order=order).fit()
        lb_p = acorr_ljungbox(fit.resid.dropna(), lags=[10], return_df=True)["lb_pvalue"].iloc[0]
        mean_results.append({
            "order": order,
            "AIC": fit.aic,
            "BIC": fit.bic,
            "LB(10) p-value": lb_p
        })
    except Exception:
        pass

mean_table = pd.DataFrame(mean_results).sort_values("AIC")
display(mean_table)

mean_model = ARIMA(Y_train, order=(2, 0, 2))
mean_fit = mean_model.fit()
print(mean_fit.summary())

mean_resid = mean_fit.resid.dropna()

arch_test = het_arch(mean_resid - mean_resid.mean(), nlags=10)
print("ARCH LM statistic:", arch_test[0])
print("ARCH LM p-value:", arch_test[1])
print("F statistic:", arch_test[2])
print("F p-value:", arch_test[3])
```

### (a) ARCH effect 檢定
為了檢驗報酬率序列是否具有 ARCH effect，先依題目要求對 `Y_t` 建立 mean model。比較多個 ARIMA 候選模型後，可選擇 `ARIMA(2,0,2)` 作為平均方程，因其 AIC 表現較佳，且殘差 Ljung-Box 檢定結果亦合理。

接著對 `ARIMA(2,0,2)` 的殘差進行 Engle 的 ARCH LM 檢定。

檢定假設如下：

- 虛無假設 `H0`：殘差不存在 ARCH effect。
- 對立假設 `H1`：殘差存在 ARCH effect。

由檢定結果可見，p-value 很小，小於 0.05，因此拒絕虛無假設，表示殘差具有顯著的 ARCH effect，適合進一步建立 GARCH 模型。

### (b) GARCH 模型建立

在確認 mean equation 的殘差存在 ARCH effect 後，可對殘差建立 `GARCH(1,1)` 模型。由於平均部分已由 `ARIMA(2,0,2)` 捕捉，因此在 GARCH 階段採用 `mean="Zero"`，只針對條件變異數建模。

本題所建立的模型可表示為：

$Y_t = \mu_t + \varepsilon_t\$

其中 $\mu_t$ 由 `ARIMA(2,0,2)` 提供，而殘差項滿足：

$\varepsilon_t = \sigma_t z_t,\quad z_t \sim N(0,1)$

$\sigma_t^2 = \omega + \alpha \varepsilon_{t-1}^2 + \beta \sigma_{t-1}^2$

此模型可用來描述金融報酬率常見的波動群聚現象。

```python
from arch import arch_model

garch_model = arch_model(mean_resid, mean="Zero", vol="GARCH", p=1, q=1, dist="normal")
garch_fit = garch_model.fit(disp="off")
print(garch_fit.summary())
```

```python
std_resid = pd.Series(garch_fit.std_resid, index=mean_resid.index).dropna()

plt.figure(figsize=(10, 4))
plt.plot(std_resid)
plt.title("標準化殘差")
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 4))
plt.plot(std_resid**2)
plt.title("標準化殘差平方")
plt.grid(True)
plt.show()

print(acorr_ljungbox(std_resid, lags=[10], return_df=True))
print(acorr_ljungbox(std_resid**2, lags=[10], return_df=True))
```

### (c) GARCH 模型殘差診斷

為檢查所建立的 GARCH 模型是否合理，可觀察標準化殘差與標準化殘差平方，並進一步進行 Ljung-Box 檢定。

- 標準化殘差的 p-value 不小，表示殘差不具有顯著自相關。
- 標準化殘差平方的 p-value 不小，表示模型已有效捕捉原本的 ARCH effect。

兩者檢定結果皆合理，便可認為此 GARCH 模型的配適效果可接受。


```python
variance_forecast = garch_fit.forecast(horizon=5).variance.iloc[-1]
volatility_forecast = np.sqrt(variance_forecast)

pd.DataFrame({
    "variance": variance_forecast.values,
    "volatility": volatility_forecast.values
}, index=[f"{i}-step" for i in range(1, 6)])
```

### (d) 波動預測

根據已建立的 `GARCH(1,1)` 模型，可在 forecast origin `2008-12-31` 下進行 1-step 至 5-step ahead 的條件變異數與波動預測。

預測結果隨步數增加而逐漸趨於平穩，表示模型呈現典型的波動均值回歸特性，這也是 GARCH 模型常見的現象。

```python
mean_forecast = mean_fit.get_forecast(steps=5).summary_frame()[["mean", "mean_ci_lower", "mean_ci_upper"]]
mean_forecast
```

### (e) 報酬率預測

本題中的 `Y_t` 預測應由 **ARIMA mean model** 提供，而不是直接用 constant mean。因為題目已明確要求先建立適當的 ARIMA 模型作為平均方程。

因此，在 forecast origin `2008-12-31` 下，可利用 `ARIMA(2,0,2)` 對未來 1-step 到 5-step 的 `Y_t` 進行預測，並同時報告其 confidence interval。

# Problem 3：Seasonal ARIMA 模型
本題使用 hsales.csv，資料內容為美國自 1973 年起的新建獨棟住宅月銷售量。由於這是一組月資料，因此很可能同時具有趨勢與季節性變動，適合使用 Seasonal ARIMA 模型進行分析。

我們的主要目標，是透過時間序列方法建立一個可以描述資料趨勢與季節結構的模型，並據此對未來的住宅銷售量進行預測。依照題目要求，需依序完成以下工作：

1. 判斷資料是否需要轉換，例如取對數，以穩定變異數。
2. 檢查資料是否平穩；若不平穩，找出合適的差分方式，包含一般差分或季節差分。
3. 辨識並選擇適合的 Seasonal ARIMA 模型。
4. 估計模型參數，並檢查殘差是否近似白噪音，以確認模型是否合理。
5. 使用最終模型預測未來 24 個月的資料。

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.stattools import adfuller

hsales_df = pd.read_csv(r"c:\Users\allen\OneDrive\Desktop\研究所\時序\hsales.csv")
hsales = hsales_df.iloc[:, -1].astype(float)
hsales.index = pd.date_range("1973-01-01", periods=len(hsales), freq="MS")

plt.figure(figsize=(10, 4))
plt.plot(hsales)
plt.title("USA 新建獨棟住宅月銷售量")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

`hsales.csv` 為美國新建獨棟住宅的月銷售量資料。由時間序列圖可觀察到資料具有明顯的季節波動，且不同時期的波動幅度不完全一致，因此適合進一步考慮轉換與季節模型。

```python
log_hsales = np.log(hsales)

plt.figure(figsize=(10, 4))
plt.plot(log_hsales)
plt.title("log(hsales)")
plt.xlabel("Date")
plt.ylabel("log sales")
plt.grid(True)
plt.show()
```

### (a) 資料是否需要轉換
由於原始資料的變異程度會隨水準改變，因此先對資料取對數轉換較為合適。對數轉換可以穩定變異數，並使季節與趨勢型態更容易分析。

```python
print("ADF p-value of raw series:", adfuller(hsales)[1])
print("ADF p-value of log series:", adfuller(log_hsales)[1])
print("ADF p-value after seasonal differencing:", adfuller(log_hsales.diff(12).dropna())[1])

seasonal_diff = log_hsales.diff(12).dropna()

plt.figure(figsize=(10, 4))
plt.plot(seasonal_diff)
plt.title("log(hsales) seasonal differencing (lag = 12)")
plt.xlabel("Date")
plt.ylabel("Differenced log sales")
plt.grid(True)
plt.show()
```

### (b) 平穩性檢查與差分
為了判斷資料是否平穩，分別對原始序列、對數轉換後序列，以及季節差分後序列進行 ADF 檢定。

檢定結果顯示：

- 原始序列的 ADF p-value 為 0.0372
- 對數轉換後序列的 ADF p-value 為 0.0191
- 經過季節差分（lag = 12）後的 p-value 為 0.0059

若僅從 ADF 檢定來看，原始序列與 log 序列已在 5% 顯著水準下拒絕單根假設，表示其不具有明顯的非季節單根。然而，從時間序列圖仍可觀察到資料具有明顯的季節性變動，因此在 Seasonal ARIMA 模型中，除了考慮一般平穩性外，也需要處理季節結構。

因此採用 季節差分（lag = 12） 作為後續建模的方式。季節差分後的序列圍繞在 0 附近波動，且 ADF p-value 進一步下降，表示該序列在處理季節性後更接近平穩，適合作為 Seasonal ARIMA 模型的分析基礎。

```python
sarima_candidates = [
    ((1,0,0), (0,1,1,12)),
    ((2,0,0), (0,1,1,12)),
    ((1,0,1), (0,1,1,12)),
    ((2,0,1), (0,1,1,12)),
    ((1,0,0), (1,1,1,12))
]

sarima_results = []
for order, seasonal_order in sarima_candidates:
    fit = SARIMAX(
        log_hsales,
        order=order,
        seasonal_order=seasonal_order,
        trend="n",
        enforce_stationarity=True,
        enforce_invertibility=True
    ).fit(disp=False)
    resid = fit.resid.dropna().iloc[24:]
    lb = acorr_ljungbox(resid, lags=[12, 24], return_df=True)
    sarima_results.append({
        "order": order,
        "seasonal_order": seasonal_order,
        "AIC": fit.aic,
        "BIC": fit.bic,
        "LB(12) p-value": lb["lb_pvalue"].iloc[0],
        "LB(24) p-value": lb["lb_pvalue"].iloc[1]
    })

pd.DataFrame(sarima_results).sort_values("AIC")
```

### (c) 模型辨識
在 Seasonal ARIMA 模型辨識階段，本研究比較數個可能的候選模型，並以 AIC、BIC 以及殘差的 Ljung-Box 檢定結果作為選模依據。由於資料為月資料，因此季節週期設定為 12。

候選模型比較結果顯示，$SARIMA(1,0,0) × (0,1,1,12)$ 的 AIC 為 -521.34，在所有候選模型中最小，表示其整體配適效果最佳。除此之外，該模型殘差的 Ljung-Box 檢定在 lag 12 與 lag 24 下的 p-value 分別為 0.5854 與 0.6203，皆大於 0.05，表示殘差中未發現顯著自相關。

因此選擇 $SARIMA(1,0,0)×(0,1,1,12)$ 作為後續分析的最終模型


```python
sarima_fit = SARIMAX(
    log_hsales,
    order=(1, 0, 0),
    seasonal_order=(0, 1, 1, 12),
    trend="n",
    enforce_stationarity=True,
    enforce_invertibility=True
).fit(disp=False)

print(sarima_fit.summary())

sarima_resid = sarima_fit.resid.dropna().iloc[24:]
print(acorr_ljungbox(sarima_resid, lags=[12, 24], return_df=True))
```

### (d) 參數估計與殘差檢定
估計結果顯示，非季節 AR(1) 係數 ar.L1 = 0.9170，其 p-value 小於 0.001，表示此參數在統計上顯著，顯示目前期的銷售量與前一期之間具有明顯的自我依賴關係。季節 MA(1) 係數 ma.S.L12 = -0.9942，雖然數值上接近 -1，但其 p-value 為 0.260，在 5% 顯著水準下未達統計顯著。即便如此，考量整體模型的 AIC 表現與殘差診斷結果，此模型仍具有不錯的配適能力。

在殘差檢定方面，Ljung-Box 檢定於 lag 12 與 lag 24 的 p-value 分別為 0.5854 與 0.6203，皆大於 0.05，表示殘差中未存在顯著自相關，可視為近似白噪音。Jarque-Bera 檢定 p-value 為 0.76，顯示殘差分配未明顯偏離常態；Heteroskedasticity 檢定 p-value 為 0.46，也未顯示顯著異質變異。

綜合以上結果，可認為 $SARIMA(1,0,0) × (0,1,1,12)$ 模型已大致捕捉資料中的主要動態與季節結構，其殘差表現合理，因此可作為後續預測的最終模型。

```python
forecast_24 = sarima_fit.get_forecast(steps=24).summary_frame()
forecast_24_level = np.exp(forecast_24[["mean", "mean_ci_lower", "mean_ci_upper"]])
forecast_24_level.index = pd.date_range(log_hsales.index[-1] + pd.offsets.MonthBegin(), periods=24, freq="MS")

forecast_24_level
```

```python
plt.figure(figsize=(12, 5))
plt.plot(hsales, label="Observed")
plt.plot(forecast_24_level.index, forecast_24_level["mean"], label="Forecast", color="red")
plt.fill_between(
    forecast_24_level.index,
    forecast_24_level["mean_ci_lower"],
    forecast_24_level["mean_ci_upper"],
    color="pink",
    alpha=0.3,
    label="95% CI"
)
plt.title("24 個月預測結果")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()
```

### (e) 未來 24 個月預測
根據前述選定的 $SARIMA(1,0,0) × (0,1,1,12)$ 模型，進一步對未來 24 個月的住宅銷售量進行預測，並將結果由 log 尺度轉回原始銷售量尺度，同時提供 95% 信賴區間。

由預測結果可見，未來兩年的銷售量大致維持在與歷史資料相近的水準，且仍呈現明顯的季節波動型態。預測值大約落在 40 至 60 左右，顯示模型預期未來住宅銷售量不會出現明顯的長期上升或下降趨勢，而是持續受到季節因素影響，在不同月份之間呈現規律性起伏。

從信賴區間來看，隨著預測期數增加，區間範圍逐漸擴大，表示長期預測的不確定性高於短期預測。這是時間序列模型常見的現象，也說明雖然模型能掌握整體趨勢與季節結構，但對較遠未來的預測仍具有一定程度的不確定性。

$SARIMA(1,0,0) × (0,1,1,12)$ 模型所得到的 24 個月預測結果合理地延續了原始資料的季節模式，因此可作為未來住宅銷售量變化的參考依據。
